```python
import pandas as pd

df = pd.read_csv("PR_fixed.csv")

residue_cols = [c for c in df.columns if c.startswith("P")]

sequence = "".join(
    df.loc[0, residue_cols]
    .dropna()
    .astype(str)
)
print(sequence)

print("Sequence length:", len(sequence))

mw = {
    'A':89,'R':174,'N':132,'D':133,'C':121,
    'E':147,'Q':146,'G':75,'H':155,'I':131,
    'L':131,'K':146,'M':149,'F':165,'P':115,
    'S':105,'T':119,'W':204,'Y':181,'V':117
}


def score_residue(r):
    return mw.get(r, 0)


k = 5
all_kmers = {l: [] for l in range(1, k+1)}


for length in range(1, k+1):
    
    for i in range(len(sequence) - length + 1):
        
        subseq = sequence[i:i+length]
        sc = sum(score_residue(r) for r in subseq)
        
        all_kmers[length].append((subseq, sc))
        
best_subseq, best_score = max(all_kmers[k], key=lambda x: x[1])

print("\nBest subseq:", best_subseq)
print("Best score:", best_score)
```
