```python
sequence = "CCTCAAATCACTCTTTGGCAACGACCCCTTGTCACAGTAAAAATAGGAGGACAGCTGAAAGAAGCTTTC"

print("Sequence length:", len(sequence))

mw = {
    'A': 89, 'R': 174, 'N': 132, 'D': 133, 'C': 121,
    'E': 147, 'Q': 146, 'G': 75, 'H': 155, 'I': 131,
    'L': 131, 'K': 146, 'M': 149, 'F': 165, 'P': 115,
    'S': 105, 'T': 119, 'W': 204, 'Y': 181, 'V': 117
}

def score_residue(r):
    return mw.get(r, 0)

max_weight = max(mw.values())

k = 5
all_kmers = []

for i in range(len(sequence) - k + 1):
    subseq = sequence[i:i+k]
    all_kmers.append(subseq)

print("Total k-mers:", len(all_kmers))
print(all_kmers)

def branch_and_bound(start, depth, subseq, score,
                     best_subseq, best_score):

    remaining = k - depth
    upper_bound = score + remaining * max_weight

    if upper_bound <= best_score:
        return best_subseq, best_score

    if depth == k:
        if score > best_score:
            return subseq, score
        return best_subseq, best_score

    for i in range(start, len(sequence)):
        r = sequence[i]
        sc = score_residue(r)

        best_subseq, best_score = branch_and_bound(
            i + 1,
            depth + 1,
            subseq + r,
            score + sc,
            best_subseq,
            best_score
        )

    return best_subseq, best_score


best_subseq, best_score = branch_and_bound(0, 0, "", 0, "", 0)

print("\nBest subsequence:", best_subseq)
print("Best score:", best_score)
```
